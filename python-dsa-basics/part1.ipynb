1. Two Sum [Hash Map]
Given a list of integers nums and an integer target, return the indices of the two numbers that add up to
the target. Assume exactly one solution exists.
Example: nums = [2, 7, 11, 15], target = 9
->
[0, 1]
Why it matters: Looking up records by key (e.g. matching user IDs to orders) is the everyday use of dicts.
Hint: Store each number's index in a dict as you loop; check if (target - n) is already stored. O(n).

```python
def two_sum(li: list[int], target: int) -> list:
    """
    input:
        li : list[int]
        target : int
    
    output: 
        [x, y] 
        where x and y are indices of the two integers where li[x] + li[y] = target
    
    Given a list of integers nums and an integer target, return the indices of the two numbers that add up to
    the target. Assume exactly one solution exists.
    Example: nums = [2, 7, 11, 15], target = 9
    ->
    [0, 1]
"""

    indices = dict()
    
    for i in range (0, len(li)):
        indices.update({li[i]: i})

        if ((target - li[i]) in indices):
            return [indices[target - li[i]], i]
        
    return []
```

```python
print(two_sum([11, 2, 15, 7], 9))
```

2. Valid Parentheses [Stack]
Given a string containing only ()[]{}, return True if every bracket is closed in the correct order.
Example: '{[()]}' -> True       
'([)]' -> False     

Why it matters: Same idea as validating nested JSON, expression parsers and template engines.
Hint: Push opening brackets; on a closing bracket, the top of the stack must be its match.

```python
def parentheses_validator(parantheses) -> bool:
    stack = []
    for p in parantheses:
        try:
            if p in ["(", "[", "{"]:
                stack.append(p)
            elif (p == ")" and "(" == stack[-1]) or (p == "}" and "{" == stack[-1]) or (p == "]" and "[" == stack[-1]):
                stack.pop()
            else:
                return False
        except:
            return False # stack maa kei element bachey last maa closing wala paranthesis ko
        
    return True  
```

```python
parentheses_validator("{[]()})")
```

```python
# efficient solution
def para_valid(parentheses) -> bool:
    pairs = {
        "}" : "{",
        ")" : "(",
        "]" : "["        
    }
    stack = []
    
    for p in parentheses:
        if p in pairs: # closing parenthesis khojdai xa
            if not stack or pairs[p] != stack.pop(): # {[]()})
                return False
        else:
            stack.append(p)
            
    return not stack # stack maa kei bachey false, (({[]()})
```

```python
para_valid("(({[]()})")
```

3. Check for Duplicates [Set]
Return True if any value appears at least twice in a list.
Example: [1, 2, 3, 1] -> True
[1, 2, 3, 4] -> False
Why it matters: Deduplicating IDs, emails or event messages before processing.
Hint: A set drops duplicates, so compare its length with the list's length.

```python
def duplicate_checker(num: list):
    if len(num) != len(set(num)):
        return True
    return False 
```

```python
duplicate_checker([1,2,3,4])
```

4. Word Frequency Counter [Dictionary]
Given a sentence, return a dictionary of how many times each word appears (ignore case).
Example: 'the cat and The dog' -> {'the': 2, 'cat': 1, 'and': 1, 'dog': 1}
Why it matters: Log analysis, counting status codes, API hits per user.
Hint: Use dict.get(word, 0) + 1, or collections.Counter.

```python
def word_freq_counter(sentence: str, case_sensitive: bool = False) -> dict:
    
    if not case_sensitive:
        sentence = str.lower(sentence)
        
    word_counter = {}
    for word in sentence.split(" "):
        if word in word_counter:
            word_counter[word] = word_counter[word] + 1
        else:
            word_counter.update({word: 1})
        
        # alternatively
        # word_counter[word] = word_counter.get(word, 0) + 1
    return word_counter
```

```python
word_freq_counter("the cat and The dog")
```
