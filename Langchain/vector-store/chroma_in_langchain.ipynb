```python
from langchain.vectorstores import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
from langchain.schema import Document
```

```python
from langchain.schema import Document

# create Langchain documents for IPL players
doc1 = Document(
    page_content="Virat Kohli is one ot the most successful and consistent batsmen in IPL history.",
    metadata={"team": "Royal Challengers Banglore"}
)
doc2 = Document(
    page_content="Rohit Sharma is the most successful captain in IPL history, leading Mumbai Indians to five titles.",
    metadata={"team": "Royal Challengers Banglore"}
)
doc3 = Document(
    page_content="MS Dhoni, famously known as Captain Cool, had led Chennai Super Kings to multiple IPL tiltes.",
    metadata={"team": "Royal Challengers Banglore"}
)
```

```python
docs = [doc1, doc2, doc3]
```

```python
vector_store = Chroma(
    embedding_function=HuggingFaceEmbeddings(),
    persist_directory="chroma_db",
    collection_name="try"
)
```

```python
vector_store.add_documents(docs) # ids are generated itself, but we can pass custome ids too
```

```python
vector_store.get(include=['embeddings', 'documents', 'metadatas'])
```

```python
# similarity search

vector_store.similarity_search(
    query="who among these won five titles?",
    k = 4, # how many similar objects you wanna see in output
)
```

```python
# search with similarity score
vector_store.similarity_search_with_score(
    query="who among these won multiple times?",
    k=4
)
```

```python
# update documents

update_doc1 = Document(
    page_content="Virat Kohli is one ot the most successful and consistent batsmen in IPL history.",
    metadata={"team": "Royal Challengers Banglore"}
)

vector_store.update_document(document_id='11ef9e9a-4a51-4d7a-932c-655f14ec6254', document=update_doc1)
```

```python
# delete document
vector_store.delete(ids=['11ef9e9a-4a51-4d7a-932c-655f14ec6254'])
```
