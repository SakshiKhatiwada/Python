```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain.vectorstores import FAISS
from langchain.schema import Document
```

```python
doc1 = Document(
    page_content="The apple is a crisp fruit that comes in varieties such as Fuji, Gala, and Granny Smith.",
    metadata={"name": "apple"}
)

doc2 = Document(
    page_content="Bananas provide natural sweetness and are rich in potassium, making them a quick energy snack",
    metadata={"name": "banana"}
)

doc3 = Document(
    page_content="A ripe mango has a fragrant aroma and a juicy, golden flesh loved across tropical countries.",
    metadata={'name': "mango"}
)

doc4 = Document(
    page_content="Oranges are citrus fruits known for their bright color, refreshing juice, and high vitamin C content",
    metadata={"name": "orange"}
)

doc5 = Document(
    page_content="Strawberries are small red berries with tiny seeds on their surface and a sweet, tart flavor",
    metadata={"name": "strawberry"}
)
```

```python
docs = [doc1, doc2, doc3, doc4, doc5]
```

```python
import faiss
from langchain_community.docstore.in_memory import InMemoryDocstore
from langchain_community.vectorstores import FAISS
```

```python
index = faiss.IndexFlatL2(len(HuggingFaceEmbeddings().embed_query("sweet fruit")))
# this len is giving just dimensions
vector_store = FAISS(
    embedding_function = HuggingFaceEmbeddings(),
    index = index,
    docstore = InMemoryDocstore(),
    index_to_docstore_id={}
)
```

```python
# adding item to vector store
vector_store.add_documents(docs)
```

```python
# delete item
vector_store.delete(ids=["53a9d7a4-4577-4010-be76-27689064623d"])
```

```python
# querying vector store

results= vector_store.similarity_search(
    "Very sweet to eat",
    k = 2
)

print(results)
```

```python
vector_store.similarity_search_with_score(
    "Very sweet to eat",
    k = 2
)
```

```python
vector_store.save_local("faiss_index")

new_vector_store = FAISS.load_local(
    'faiss_index', HuggingFaceEmbeddings(), allow_dangerous_deserialization=True
)

docs = new_vector_store.similarity_search("citric")
```
