## Standardizing these two classes: FakeLLM and FakePromptTemplate

```python
import random
```

```python
# we will create an abstract class called Runnable
from abc import ABC, abstractmethod

class Runnable(ABC):
    @abstractmethod
    def invoke(input_data):
        pass
```

```python
class FakeLLM(Runnable):
    def __init__(self):
        print("LLM created")

    def invoke(self, prompt):
        response_list = [
            "Kathmandu is the capital of Nepal",
            "IPL is a cricket league",
            "AI stands for Artificial Intelligence",
        ]
        return {"response": random.choice(response_list)}

    def predict(
        self, prompt
    ):  # not removing completely, a msg should be printed for deprication
        response_list = [
            "Kathmandu is the capital of Nepal",
            "IPL is a cricket league",
            "AI stands for Artificial Intelligence",
        ]
        print("This method is going to be depricated. Encouraged to use invoke method.")
        return {"response": random.choice(response_list)}
```

```python
class FakeStrOutputParser(Runnable):
    def __init__(self):
        pass
    
    def invoke(self, input_data):
        return input_data['response']
```

```python
llm = FakeLLM()
```

```python
class FakePromptTemplate:
    def __init__(self, template, input_variables):
        self.template = template
        self.input_variables = input_variables
        
    def invoke(self, input_dict): 
        return self.template.format(**input_dict)
    
    def format(self, input_dict): 
        print("This method is about to be depricated. Use Invoke.")
        return self.template.format(**input_dict)
```

```python
template = FakePromptTemplate(
    template='Describe {topic} in {length}',
    input_variables=['topic', 'length']
)
```

```python
class RunnableConnector(Runnable):
    def __init__(self, runnable_list):
        self.runnable_list = runnable_list
        
    def invoke(self, input_data):
        # main code
        for runnable in self.runnable_list:
            input_data = runnable.invoke(input_data) 
        return input_data 
```

```python
parser = FakeStrOutputParser()
```

```python
# chain = RunnableConnector([prompt, llm])
chain = RunnableConnector([template, llm, parser])
```
