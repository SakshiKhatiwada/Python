```python
import random
class FakeLLM:
    def __init__(self):
        print("LLM created")
    
    def predict(self, prompt):
        response_list = [
            "Kathmandu is the capital of Nepal",
            "IPL is a cricket league",
            "AI stands for Artificial Intelligence"
        ]
        
        return {'response': random.choice(response_list)}
```

```python
llm = FakeLLM()
```

```python
llm.predict("What is AI?")
```

```python
class FakePromptTemplate:
    def __init__(self, template, input_variables):
        self.template = template
        self.input_variables = input_variables
        
    def format(self, input_dict): # this is a function "format"
        return self.template.format(**input_dict) # this is a string formatting function. Both are different
```

```python
template = FakePromptTemplate(
    template='Describe {topic} in {length}',
    input_variables=['topic', 'length']
)
```

```python
prompt = template.format({'topic': 'penguin', 'length': "short"})
```

```python
llm = FakeLLM()
llm.predict(prompt)
```

```python
class FakeLLMChain:
    def __init__(self, llm, prompt):
        self.llm = llm
        self.prompt = prompt

    def run(self, input_dict):
        final_prompt = self.prompt.format(input_dict)  # format function we created
        result = self.llm.predict(final_prompt)
        return result["response"]
```

```python
template = FakePromptTemplate(
    template='Describe {topic} in {length}',
    input_variables=['topic', 'length']
)
llm = FakeLLM()
```

```python
chain = FakeLLMChain(llm, template)
```

```python
chain.run({"length": "short", "topic": "nepal"})
```
