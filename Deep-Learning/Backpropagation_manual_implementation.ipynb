```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame([[8,8,4],[7,9,5], [6,10,6],[5,12,7]], columns=['cgpa', 'profile_score', 'lpa'])
```

```python
df
```

```python
def initialization_parameters(layer_dim):
  parameters = {}
  np.random.seed(3)
  L = len(layer_dim)
  for l in range(1,L):
    parameters['W'+str(l)] = np.ones((layer_dim[l-1],layer_dim[l]))* 0.1
    parameters['b'+str(l)] = np.zeros((layer_dim[l], 1))
  return parameters
```

```python
initialization_parameters([2,2,1])
```

```python
def linear_forward(A_prev, W,b):
  Z = np.dot(W.T, A_prev) + b
  return Z
```

```python
# Forward Propagation

def L_layer_forward(X, parameters):
  A = X
  L = len(parameters) // 2 # => no. of layers in NN

  for l in range(1, L+1):
    A_prev = A
    Wl = parameters['W' + str(l)]
    bl = parameters['b' + str(l)]

    # print("A" + str(l-1)+ ": ", A_prev)
    # print("W" + str(l) + ": ", Wl)
    # print("b" + str(l) + ": ", bl)
    # print("--" * 20)

    A = linear_forward(A_prev, Wl, bl)
    # print("A" + str(l) + ": ", A)
    # print("**" * 20)

  return A,A_prev
```

```python
X = df[['cgpa', 'profile_score']].values[0].reshape(2,1) # shape (no. of features, no. of training example)
y = df[['lpa']].values[0][0]
```

```python
df[['lpa']]
```

```python
# Parameter Initialization

parameters = initialization_parameters([2,2,1])
parameters
```

```python
y_hat, A1 = L_layer_forward(X, parameters)
```

```python
y_hat = y_hat[0,0]
```

```python
A1
```

```python
# loss
(y - 0.32) ** 2
```

```python
def update_parameters(parameters, y, y_hat, A1, X):
  temp = 0.001 * 2 * (y-y_hat)
  parameters['W2'][0][0] += ( temp * A1[0][0])
  parameters['W2'][1][0] += ( temp  * A1[1][0])
  parameters['b2'][0][0] += ( temp )

  parameters['W1'][0][0] += (temp * parameters['W2'][0][0] * X[0][0])
  parameters['W1'][0][1] += (temp * parameters['W2'][0][0] * X[1][0])
  parameters['b1'][0][0] += (temp * parameters['W2'][0][0])

  parameters['W1'][1][0] += (temp * parameters['W2'][1][0] * X[0][0])
  parameters['W1'][1][1] += (temp * parameters['W2'][1][0] * X[1][0])
  parameters['b1'][1][0] += (temp * parameters['W2'][1][0])
```

```python
update_parameters(parameters, y, y_hat, A1, X)
```

```python
# and continueeeeeeee...


# Implementing all in one: Epochs Implementation

parameters = initialization_parameters([2,2,1])
epochs = 3

for i in range(epochs):
  loss = []

  for j in range(df.shape[0]):
    X = df[['cgpa', 'profile_score']].values[j].reshape(2,1)
    # shape (no. of features, no. of training example)
    y = df[['lpa']].values[j][0]

    # parameter initialization

    y_hat, A1 = L_layer_forward(X, parameters)
    y_hat = y_hat[0,0]

    update_parameters(parameters, y, y_hat, A1, X)

    loss.append((y-y_hat)**2)

  print("Epoch-", i+1, 'loss: ', np.array(loss).mean())

parameters
```
