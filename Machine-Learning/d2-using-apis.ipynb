```python
import os
```

```python
TMDB_API_ACCESS_TOKEN = os.getenv("TMDB_API_ACCESS_TOKEN")
```

```python
import requests

url = "https://api.themoviedb.org/3/movie/top_rated?language=en-US&page=1"

headers = {
            "accept": "application/json",
           'Authorization': f"Bearer {TMDB_API_ACCESS_TOKEN}"
}

response = requests.get(url, headers=headers)

if response.status_code == 200:
    print(response.json())
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(response.json()['results'])
```

```python
df.head()
```

```python
df = df[['id', 'title', 'overview','release_date', 'popularity', 'vote_average', 'vote_count']]
```

```python
df.shape # 1 paage, 20 movies
```

```python
# now we will do that in loop for all pages

df = pd.DataFrame()

for i in range(1, 429):
    url = f"https://api.themoviedb.org/3/movie/top_rated?language=en-US&page={i}"

    headers = {
            "accept": "application/json",
           'Authorization': f"Bearer {TMDB_API_ACCESS_TOKEN}"
    }

    response = requests.get(url, headers=headers)
    df= pd.concat([df, pd.DataFrame(response.json()['results'])[['id', 'title', 'overview','release_date', 'popularity', 'vote_average', 'vote_count']]])
```

```python
df.reset_index(drop=True, inplace=True)
```

```python
# same code but only 1 time fetching from the server => 2.3s compared to 4m 28s in the above code

df = pd.DataFrame()

url = f"https://api.themoviedb.org/3/movie/top_rated"
headers = {
            "accept": "application/json",
           'Authorization': f"Bearer {TMDB_API_ACCESS_TOKEN}"
    }
response = requests.get(url, headers=headers)

for i in range(1, response.json()['total_pages'] + 1):

    df= pd.concat([df, pd.DataFrame(response.json()['results'])[['id', 'title', 'overview','release_date', 'popularity', 'vote_average', 'vote_count']]], ignore_index=True)
```
