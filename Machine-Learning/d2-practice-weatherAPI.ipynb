```python
import pandas as pd
import requests
import os
```

```python
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
RAPID_API_KEY = os.getenv("RAPID_API_KEY")
RAPID_API_KEY
```

```python
url = "https://weatherbit-v1-mashape.p.rapidapi.com/forecast/3hourly"

querystring = {"lat":"28","lon":"84","units":"metric","lang":"en"}

headers = {
	"x-rapidapi-key": RAPID_API_KEY,
	"x-rapidapi-host": "weatherbit-v1-mashape.p.rapidapi.com"
}

response = requests.get(url, headers=headers, params=querystring)

response.json()
```

```python
df = pd.DataFrame(response.json())
```

```python
df
```

```python
df['weather_description'] = df['data'][1]['weather']['description']
df['timestamp_local'] = df['data'][1]['timestamp_local']
df['timestamp_utc'] = df['data'][1]['timestamp_utc']
df['datetime'] = df['data'][1]['datetime']
```

```python
df.drop(columns={"data"},inplace=True)
```
