## Checking the Speed

```python
import time

size = 10000

# create random matrices on CPU
matrix_cpu1 = torch.randn(size, size) # draws sample from normal distribution
matrix_cpu2 = torch.randn(size, size)
```

```python
print(matrix_cpu1[1])
```

```python
# Measuring time on CPU
start_time = time.time()
result_cpu = torch.matmul(matrix_cpu1, matrix_cpu2)
cpu_time = time.time() - start_time
print("CPU time: ", cpu_time)
```

```python
# moving to GPU
matrix_gpu1 = matrix_cpu1.to("cuda")
matrix_gpu2 = matrix_cpu2.to("cuda")

# measuring time on GPU
start_time = time.time()
result_gpu = torch.matmul(matrix_gpu1, matrix_gpu2)
torch.cuda.synchronize() # ensures all GPU are complete
gpu_time = time.time() - start_time
print("GPU time: ", gpu_time)
```

```python
# comparing results
print("Speedup (CPU time/GPU time): ", cpu_time/gpu_time)
```
