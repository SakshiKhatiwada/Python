## Creating a Tensor

```python
# using empty
a = torch.empty(3,5) # memory allocation, doesn't initialize
a
```

```python
type(a)
```

```python
# using zeros
torch.zeros(2,3)
```

```python
# using ones

torch.ones(1,2)
```

```python
# using rand

torch.rand(2,3) # initialized between 0 and 1
```

```python
# use of manual_seed
torch.manual_seed(100)
print(torch.rand(2,3))
torch.manual_seed(100)
torch.rand(2,3)
```

```python
# custom tensor
torch.tensor([[1,2,3],[4,5,6]])
```

```python
# other ways

# arange
print("using arange -> ", torch.arange(0,10,2)) # start, end, step

# linspace
print("using linspace: ", torch.linspace(0,10,10))

# eye
print("using eye: ", torch.eye(5))

# full
print("using full: \n", torch.full((3,3), 5))
```

## Mathematical Operations

```python
a = torch.tensor([1,2,3], dtype=torch.int32)
print(a + 2)
print(a - 3)
print(a * 4)
print(a / 2 )
print(a * 5)
print(a//2)
print(a % 2)
print('multiply: ',a * 5)
print(a**2)
```

```python
# element-wise operation
a = torch.rand(2,3)
b = torch.rand(2,3)

print(a,"\n", b)
a + b # element-wise operations
```

```python
# abs
torch.abs(torch.tensor([1,-2,3]))
```

```python
torch.neg(torch.tensor([1,-2,3]))
```

```python
d = torch.tensor([1,2.3, 4.1, 5.5])
torch.round(d)
```

```python
print("ceiling: ", torch.ceil(d))
print("floor: ", torch.floor(d))
```

```python
# clamp
torch.clamp(d, min=2, max=4)
```

## Reduction Operation

```python
e = torch.randint(size=(2,3), low=0, high=10)
e
```

```python
# sum
torch.sum(e)
```

```python
# sum along columns and rows
torch.sum(e, dim=0) # col wise
torch.sum(e, dim=1) # row wise
```

```python
# mean
# torch.mean(e) # only works with float, not long
f = e.to(dtype=torch.float32)
torch.mean(f)
```

```python
# mean along rows and columns
torch.mean(f, dim=0)
```

```python
# median
torch.median(f)
```

```python
# max and min
print(torch.max(f))
torch.min(f)
```

```python
torch.prod(e)
```

```python
torch.std(f)
```

```python
# variance
torch.var(f)
```
