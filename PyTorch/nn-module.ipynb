```python
import torch
import torch.nn as nn
```

```python
class Model(nn.Module):
    def __init__(self, num_features):
        super().__init__()
        # self.linear = nn.Linear(num_features, 1) # input features and output features
        # self.sigmoid = nn.Sigmoid()
        
        #---------------- Improved Code
        self.linear1 = nn.Linear(num_features, 3) # input features and output features
        self.relu = nn.ReLU()
        self.linear2 = nn.Linear(3,1)
        self.sigmoid = nn.Sigmoid()
        
    def forward(self, features):
        # out = self.linear(features)
        # out = self.sigmoid(out)
        
        #---------------- Improved Code
        out = self.linear1(features)
        out = self.relu(out)
        out = self.linear2(out)
        out = self.sigmoid(out)
        return out
```

```python
# dataset creation
features = torch.rand(10,5)

# model creation
model = Model(features.shape[1])

# call model for forward pass
model(features) # same as model.forward(features), in a standard way. This is a magic method
```

```python
# show model's weights
print(model.linear1.bias)
model.linear1.weight
```

```python
# Sequential Container

class Model2(nn.Module):
    def __init__(self, num_features):
        super().__init__()
        self.network = nn.Sequential(
            nn.Linear(num_features, 3), 
            nn.ReLU(),
            nn.Linear(3,1),
            nn.Sigmoid()
        )
        
    def forward(self, features):
        out = self.network(features)
        return out
```

```python
# dataset creation
features = torch.rand(10,5)

# model creation
model = Model2(features.shape[1])

# call model for forward pass
model(features) # same as model.forward(features), in a standard way. This is a magic method
```
