```python
# create model class
import torch
import torch.nn as nn


class Model(nn.Module): # inheriting from nn.Module class
    def __init__(self, num_features):
        super().__init__()
        
        # self.linear = nn.Linear(num_features, 1)
        # self.sigmoid = nn.Sigmoid()
        
        # self.linear1 = nn.Linear(num_features, 3)
        # self.relu = nn.ReLU()
        # self.linear2 = nn.Linear(3,1)
        # self.sigmoid = nn.Sigmoid()
        
        self.network = nn.Sequential( # sequential container
            nn.Linear(num_features,3),
            nn.ReLU(),
            nn.Linear(3,1),
            nn.Sigmoid()
        )
        
    def forward(self, features):
        # out = self.linear1(features)
        # out = self.relu(out)
        # out = self.linear2(out)
        # out = self.sigmoid(out)
        
        out = self.network(features)
    
        return out
    
    
```

```python
# create dataset
features = torch.rand(10,5)

# create model
model = Model(features.shape[1])

# call model for forward pass
# model.forward(features) # we can do that but it's not recommended
model(features) # forward method is triggered itself *IMP - Magic Method
```
